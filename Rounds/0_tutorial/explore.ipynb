```python
spread = df_t['ask_price_1'].values - df_t['bid_price_1'].values

print(spread)


mean_spread = spread[1].mean()

print(mean_spread)

```

```python
plt.figure(figsize=(10, 5))
plt.hist(df_t['ask_volume_1'].values, bins=30, color='mediumpurple', edgecolor='black', alpha=0.7)
plt.hist(df_t['bid_volume_1'].values, bins=30, color='red', edgecolor='black', alpha=0.7)
#plt.axvline(mean_spread, color='lime', linewidth=1.5, linestyle='--', label=f'Mean = {mean_spread:.2f}')

plt.title('Emeralds Bid-Ask Spread Histogram')
plt.xlabel('Spread (ask - bid)')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```
