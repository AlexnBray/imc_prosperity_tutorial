```python
import sys
import plotly.graph_objects as go
from pathlib import Path
import pandas as pd
import numpy as np
import re
# --- Data Loading ---

files = [
    'prices_round_5_day_2.csv',
    'prices_round_5_day_3.csv',
    'prices_round_5_day_4.csv'
]

prices = []
for file in files:
    try:
        df = pd.read_csv(file, sep=';')
        day_match = re.search(r'day_(-?\d+)', file)
        if day_match:
            df['day'] = int(day_match.group(1))
        prices.append(df)
    except FileNotFoundError:
        print(f"Warning: {file} not found.")

df_total = pd.concat(prices, ignore_index=True)
df_total = df_total.sort_values(by=['day', 'timestamp']).reset_index(drop=True)
```

```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np

# 1. Filter and Prepare Data
products = ["SNACKPACK_CHOCOLATE", "SNACKPACK_VANILLA"]
target_df = df_total[df_total["product"].isin(products)].copy()

# Pivot for the specific metrics requested
pivoted_bid = target_df.pivot_table(index=["day", "timestamp"], columns="product", values="bid_price_1")
pivoted_ask = target_df.pivot_table(index=["day", "timestamp"], columns="product", values="ask_price_1")
pivoted_mid = target_df.pivot_table(index=["day", "timestamp"], columns="product", values="mid_price")

# --- CALCULATIONS ---

# Sums and Averages for Prices
sum_bid = pivoted_bid.sum(axis=1)
sum_ask = pivoted_ask.sum(axis=1)
mid_v = pivoted_mid["SNACKPACK_VANILLA"]
mid_c = pivoted_mid["SNACKPACK_CHOCOLATE"]
avg_of_mids = (mid_v + mid_c) / 2

# Long / short chunked medians + SD snap toward short when divergence is extreme
# Per day: long level = median(mid) refreshed every LONG ticks (flat steps); short same with SHORT ticks.
# σ = rolling stdev(mid − long) over SIGMA_WINDOW; use short stepping mean iff |short−long| > K_SD × max(σ, floor).


def _chunked_stepped_median(m: np.ndarray, period: int) -> np.ndarray:
    """Piecewise-flat series: median(mid) on each contiguous chunk of length `period`."""
    n = len(m)
    out = np.empty(n, dtype=float)
    period = max(1, int(period))
    for start in range(0, n, period):
        end = min(n, start + period)
        out[start:end] = float(np.nanmedian(m[start:end]))
    return out


def long_short_snap(mid: pd.Series, long_ticks: int, short_ticks: int, k_sd: float, sigma_window: int, sigma_floor: float):
    snaps, longs = [], []
    sigma_window = max(2, int(sigma_window))
    min_periods = max(50, min(sigma_window // 10, sigma_window))

    for _day, grp in mid.groupby(level="day", sort=True):
        m = grp.astype(float).to_numpy()
        mu_l = _chunked_stepped_median(m, long_ticks)
        mu_s = _chunked_stepped_median(m, short_ticks)
        resid = m - mu_l
        sig = pd.Series(resid).rolling(sigma_window, min_periods=min_periods).std().to_numpy()
        thresh = np.where(np.isnan(sig), sigma_floor, np.maximum(sig, sigma_floor))
        divergence = np.abs(mu_s - mu_l)
        follow_short = divergence > k_sd * thresh
        snap = np.where(follow_short, mu_s, mu_l)
        snaps.append(pd.Series(snap, index=grp.index))
        longs.append(pd.Series(mu_l, index=grp.index))

    return pd.concat(snaps), pd.concat(longs)


LONG_STEP_TICKS = 800
SHORT_STEP_TICKS = 80
K_SD_THRESHOLD = 8
SIGMA_WINDOW_TICKS = 500
SIGMA_FLOOR = 1e-3  # avoids σ≈0 over-triggering snaps

mid_v_snap, mid_v_long = long_short_snap(
    mid_v,
    LONG_STEP_TICKS,
    SHORT_STEP_TICKS,
    K_SD_THRESHOLD,
    SIGMA_WINDOW_TICKS,
    SIGMA_FLOOR,
)
mid_c_snap, mid_c_long = long_short_snap(
    mid_c,
    LONG_STEP_TICKS,
    SHORT_STEP_TICKS,
    K_SD_THRESHOLD,
    SIGMA_WINDOW_TICKS,
    SIGMA_FLOOR,
)


# NEW: Calculate Cumulative Returns per Day
# 1. Get raw periodic returns
returns_v = pivoted_mid.groupby(level='day')["SNACKPACK_VANILLA"].pct_change().fillna(0)
returns_c = pivoted_mid.groupby(level='day')["SNACKPACK_CHOCOLATE"].pct_change().fillna(0)

# 2. Convert to Cumulative Returns: (1 + r1)(1 + r2)... - 1
cum_ret_v = (1 + returns_v).groupby(level='day').cumprod() - 1
cum_ret_c = (1 + returns_c).groupby(level='day').cumprod() - 1

# 3. Calculate Average Cumulative Return and its Residuals
avg_cum_ret = (cum_ret_v + cum_ret_c) / 2
resid_cum_v = cum_ret_v - avg_cum_ret
resid_cum_c = cum_ret_c - avg_cum_ret
abs_resid_cum = abs(resid_cum_v) + abs(resid_cum_c)

# Create consistent time labels
time_labels = (
    pivoted_mid.index.get_level_values("day").astype(str)
    + " / "
    + pivoted_mid.index.get_level_values("timestamp").astype(str)
)

# --- GRAPH 1: Aggregate Bid/Ask Sums ---
fig1 = go.Figure()
fig1.add_trace(go.Scatter(x=time_labels, y=sum_bid, name="Σ Bid", line=dict(color="green", width=1.5)))
fig1.add_trace(go.Scatter(x=time_labels, y=sum_ask, name="Σ Ask", line=dict(color="red", width=1.5)))
fig1.update_layout(title="Aggregate Basket Liquidity", xaxis_title="Time", yaxis_title="Price Sum", template="plotly_white")
fig1.show()

# --- GRAPH 2: Mid Price Comparison ---
fig2 = go.Figure()
fig2.add_trace(go.Scatter(x=time_labels, y=mid_v, name="Vanilla Mid", line=dict(color="orange", dash="dot")))
fig2.add_trace(go.Scatter(x=time_labels, y=mid_c, name="Chocolate Mid", line=dict(color="brown", dash="dot")))
fig2.add_trace(go.Scatter(x=time_labels, y=avg_of_mids, name="Basket Average Mid", line=dict(color="black", width=2)))
fig2.add_trace(
    go.Scatter(
        x=time_labels,
        y=mid_v_long,
        name=f"Vanilla long step (median / {LONG_STEP_TICKS} tk)",
        line=dict(color="#fd8d3c", width=1.2, dash="dash"),
        opacity=0.55,
    )
)
fig2.add_trace(
    go.Scatter(
        x=time_labels,
        y=mid_v_snap,
        name=f"Vanilla snap if |Δ| > {K_SD_THRESHOLD}·σ ({SHORT_STEP_TICKS} tk short)",
        line=dict(color="#fd8d3c", width=2.5),
    )
)
fig2.add_trace(
    go.Scatter(
        x=time_labels,
        y=mid_c_long,
        name=f"Chocolate long step (median / {LONG_STEP_TICKS} tk)",
        line=dict(color="#daa520", width=1.2, dash="dash"),
        opacity=0.55,
    )
)
fig2.add_trace(
    go.Scatter(
        x=time_labels,
        y=mid_c_snap,
        name=f"Chocolate snap if |Δ| > {K_SD_THRESHOLD}·σ ({SHORT_STEP_TICKS} tk short)",
        line=dict(color="#5d2e0f", width=2.5),
    )
)
fig2.update_layout(title="Mid Price Comparison", xaxis_title="Time", yaxis_title="Price", template="plotly_white")
fig2.show()

# --- GRAPH 3: Cumulative Return Comparison ---
fig3 = go.Figure()
fig3.add_trace(go.Scatter(x=time_labels, y=cum_ret_v, name="Vanilla Cum Return", line=dict(color="orange", width=1)))
fig3.add_trace(go.Scatter(x=time_labels, y=cum_ret_c, name="Chocolate Cum Return", line=dict(color="brown", width=1)))
fig3.add_trace(go.Scatter(x=time_labels, y=avg_cum_ret, name="Avg Basket Cum Return", line=dict(color="black", width=2)))
fig3.update_layout(
    title="Intraday Cumulative Returns (Resets Daily)",
    xaxis_title="Time", 
    yaxis_title="Cumulative Return (%)", 
    template="plotly_white",
    yaxis_tickformat=".2%" # Formats the axis as percentages
)
fig3.show()

# --- GRAPH 4: Residuals of Cumulative Returns ---
fig4 = go.Figure()
fig4.add_trace(go.Scatter(x=time_labels, y=resid_cum_v, name="Vanilla Cum Residual", line=dict(color="orange", dash="dot")))
fig4.add_trace(go.Scatter(x=time_labels, y=resid_cum_c, name="Chocolate Cum Residual", line=dict(color="brown", dash="dot")))
fig4.add_trace(go.Scatter(x=time_labels, y=abs_resid_cum, name="Total ABS Cum Residual", line=dict(color="purple", width=2)))
fig4.update_layout(
    title="Residuals of Cumulative Returns (Divergence from Basket Average)",
    xaxis_title="Time",
    yaxis_title="Residual (%)",
    template="plotly_white",
    hovermode="x unified",
    yaxis_tickformat=".2%"
)
fig4.show()
```

```python
# --- CALCULATIONS FOR STEPPING REGIME SHIFT ---

# 1. Calculate the raw spread
raw_spread = mid_v - mid_c

# 2. Calculate the Stepping Mean (Median of Medians)
# Pass 1: Rolling median to filter out spikes (Fast)
fast_median = raw_spread.groupby(level='day').rolling(window=500).median().reset_index(0, drop=True)

# Pass 2: Rolling median of the fast median to create the "Step" (Slow)
# Increasing this window makes the "steps" longer and more stable
stepping_mean = fast_median.groupby(level='day').rolling(window=1000).median().reset_index(0, drop=True)

# 3. Calculate the divergence from the step
spread_divergence = raw_spread - stepping_mean

# --- GRAPH 5: Stepping Mean Analysis ---
fig5 = go.Figure()

# Plot the raw spread (Low opacity to make the step stand out)
fig5.add_trace(go.Scatter(
    x=time_labels, 
    y=raw_spread, 
    name="Raw Spread (V - C)", 
    line=dict(color="rgba(128, 128, 128, 0.4)", width=1)
))

# Plot the stepping mean - This is your "Parallel Mean"
fig5.add_trace(go.Scatter(
    x=time_labels, 
    y=stepping_mean, 
    name="Stepping Mean (Regime)", 
    line=dict(color="blue", width=3, shape='hv') # 'hv' ensures horizontal-vertical drawing
))

# Highlight the divergence as a bar chart
fig5.add_trace(go.Bar(
    x=time_labels, 
    y=spread_divergence, 
    name="Divergence from Step",
    marker_color="red",
    opacity=0.6
))

fig5.update_layout(
    title="Spread Regime Analysis: Stepping Mean Identification",
    xaxis_title="Time",
    yaxis_title="Price Spread ($)",
    template="plotly_white",
    hovermode="x unified"
)
fig5.show()
```
